# 随机tensor

```python
a = torch.rand(4,3)   #随机生成，服从均匀分布，范围在0到1之间
print(a)
b = torch.randn(2,2)    #随机生成，服从正态分布，均值为0，标准差为1
print(b)
```

```python
a = torch.normal(mean = 0.0 , std = torch.rand(5))
#自定义均值、标准差的正态分布
#mean是标量，std是张量：均值全部相同，每个元素标准差不同
print(a)
print(a.type())
```

```python
mean = torch.tensor([1.,2.,3.,4.,5.])
a = torch.normal(mean = mean , std = 2.)
#mean是张量，std是标量：每个元素均值不同，标准差全部相同
print(a)
print(a.type())
```

```python
a = torch.normal(mean = 2. , std = 1. , size = (5,))
#mean,std都是标量，需要size输出形状
print(a)
```

```python
mean = torch.tensor([2.,1.,2.,1.])
std = torch.tensor([1.,2.,3.,4.])
a = torch.normal(mean = mean , std = std)
#mean,std都是张量，要保证mean和std两个张量的shape必须相同
print(a)
```

```python
a = torch.zeros(4,3)
a.uniform_(-1.,3.)
#均匀分布，可以自定义上下限，torch.uniform_(low,high)
print(a)
b = torch.zeros(4,3).uniform_(-1.,3.)
print(b)
```

```python
a = torch.arange(0,10,2)    #左闭右开，步长为2的序列
print(a)
```

## 稀疏的张量

```python
#indices是非零元素的坐标
indices = torch.tensor([[0,1,1],
                        [2,0,2]])
#values对应非零元素的值
values = torch.tensor([3,4,5],dtype=torch.float32)
#构造稀疏coo张量
x = torch.sparse_coo_tensor(indices, values,[2,4])
print(x)
print(x.to_dense())     #还原回稠密矩阵
```

## tensor的算术运算

```python
#加法
a = torch.rand(2,3)
b = torch.randn(2,3)
print(a)
print(b)
```

```python
print(a + b)
```

```python
#乘法（除法同理）
x = torch.tensor([[1,2,3],
                 [4,5,6]])
y = torch.tensor([[2,3,4],
                 [5,6,7]])
print(x * y)
```

```python
print(torch.mul(x,y))
```

```python
print(x.mul(y))
```

```python
print(x.mul_(y))
print(x)
```

```python
#矩阵乘法
x = torch.ones(2,1)
y = torch.ones(1,2)
print(x @ y)
```

```python
print(x.matmul(y))
```

```python
print(torch.matmul(x,y))
```

```python
print(torch.mm(x,y))    #torch.mm()只支持2维矩阵，不支持更高维度
```

```python
print(x.mm(y))
```

```python
#高维度矩阵乘法
x = torch.ones(1,2,3,4)
y = torch.ones(4,2,4,3)
print(x @ y)
#x @ y只对除了最后两维之外的前面所有维度做广播；最后两维是矩阵，不广播，要满足矩阵乘法规则
#广播机制的规则：右对齐+逐维比较，两个张量在某个维度上的大小相同，则该维度保留，大小不变；
#若其中一个为1，则该维度被扩展为另一个的大小；若两者都不为1且不相等，则无法广播
#广播后的结果，相乘得到最后有几个矩阵，该题中广播后的为(4,2)，则得到8个矩阵
print(x.matmul(y).shape)
```

## 比较运算

```python
import torch
a = torch.rand(2,2)
b = torch.rand(2,2)
print(a)
print(b)
```

```python
print(torch.gt(a,b))    #判断a是否大于b
```

```python
print(torch.ge(a,b))    #判断a是否大于等于b
```

```python
print(torch.lt(a,b))    #判断a是否小于b
```

```python
print(torch.le(a,b))    #判断a是否小于等于b
```

```python
print(torch.eq(a,b))    #判断a是否等于b，逐元素对比然后返回值
print(torch.equal(a,b))     #判断a是否等于b，整体对比返回一个值
```

```python
print(torch.ne(a,b))    #判断a是不是不等于b
```

## 排序和取值

```python
import torch
x = torch.tensor([
                [[3,2,3,5],
                 [4,5,1,9],
                  [1,2,3,4]],

                [[1,2,4,5],
                 [2,3,5,1],
                 [3,4,5,2]]
                 ])
val , idx = torch.sort(x , dim = 0 , descending = True)
#根据shape得到这是个三维的张量，dim=0，则固定除0维以外的所有维度，在每列内部，沿着第0维的方向排序，该张量中则按照第一个矩阵和第二个矩阵作对比，各个位置元素进行比较排序
#在不设置dim时，默认dim = -1，即最后一个维度，descending为true时为降序排序
print(x.shape)
print(val)
```

## 范数运算

```python
import numpy as np
data = np.array([1,2,45,6])
x_min = data.min()
x_max = data.max()
norm = (data - x_min) / (x_max - x_min)
print(norm)

miu = data.mean()
sigma = data.std()
std = (data - miu) / sigma
print(std)
```
